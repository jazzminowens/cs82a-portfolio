# Migration Note: Google Colab to JupyterHub

## What Transfers Unchanged
I would first need to download the Colab notebook as .ipynb. The file will then need to be uploaded to the JupyterHub.
My Python code and markdowns from my Colab notebook would transfer unchanged. Also, some of my saved outputs depending on what Python code was used.

## What Needs Redoing
I would need to reload the csv files to JupyterHub and remove any code that Colab used to upload files to Google drive.
Also, I would need to redo any Colab interactive form fields. Although Colab preloads libraries like pandas and seaborn, I will need to verify which Python libraries are installed on the JupyterHub environment. If the libraires needed are not installed I would need to reinstall them. I would also need to get permission to sign into the campus JupyterHub, which may require checking if I have access through my student login.

## One Risk
One risk of migrating from Colab to JupyterHub is that my notebook may not run correctly. because of missing files or Python packages not being installed. For example, if I had a CSV stored on google drive this may return a file not found error when I run it on JupyterHub. 

## Experience Migrating from local  Jupyter to Colab
I have some experience migrating from local Jupyter to Colab. Although I was able to use the same Python code, I did get some errors when running the code. For example, since I uploaded a CSV file to Jupyter when migrated to  Colab there was a File not found error. Also, I had to upload the csv file again on Colab when I reopened the notebook. 

## Conclusion
Although everything may not transfer when migrating from different platforms it is still a good skill to have. It doesn't require me to start my assignments completely over but may require a few extra steps like reloading files and libraries.
