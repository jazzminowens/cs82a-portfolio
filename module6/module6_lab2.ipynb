```python
import pandas as pd
import sqlite3
con = sqlite3.connect("lab2.db")
```

```python
con.execute("""
CREATE TABLE IF NOT EXISTS customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT,
    age INTEGER
)
""")
```

```python
con.execute("DELETE FROM customers")
con.commit()
```

```python
con.execute("""
INSERT INTO customers VALUES
(1, 'Tracy Owens', 'Los Angeles', 35),
(2, 'Melody Davis', 'Detroit', 26),
(3, 'James Jones', 'New York', 39),
(4, 'Nick Taylor', 'Houston', 42),
(5, 'Sammy Johnson', 'Dallas', 56),
(6, 'Travis Barker', 'San Diego', 21),
(7, 'Chris Ross', 'Miami', 38),
(8, 'Walter Reed', 'Las Vegas', 45)
""")
con.commit()
```

```python
pd.read_sql("SELECT * FROM customers", con)
```

The difference between MySQL-to- SQLite is that MySQL uses int(11) and char(35), while SQLite uses INTEGER and TEXT.
