```python
import pandas as pd
import sqlite3
con = sqlite3.connect("lab3.db")
```

```python
con.execute("""
CREATE TABLE IF NOT EXISTS customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    city TEXT,
    age INTEGER
)
""")
```

```python
con.execute("DELETE FROM customers")
con.commit()
```

```python
con.execute("""
INSERT INTO customers VALUES
(1, 'Tracy Owens', 'Los Angeles', 35),
(2, 'Melody Davis', 'Detroit', 26),
(3, 'James Jones', 'New York', 39),
(4, 'Nick Taylor', 'Houston', 42),
(5, 'Sammy Johnson', 'Dallas', 56),
(6, 'Travis Barker', 'San Diego', 21),
(7, 'Chris Ross', 'Miami', 38),
(8, 'Walter Reed', 'Las Vegas', 45)
""")
con.commit()
```

```python
pd.read_sql("SELECT * FROM customers", con)
```

```python
pd.read_sql("SELECT * FROM customers WHERE age> 30", con)
```

```python
pd.read_sql("SELECT * FROM customers ORDER BY age ASC", con)
```

```python
pd.read_sql("SELECT * FROM customers WHERE name LIKE '%ic%'", con)
```

```python
df_age = pd.read_sql("SELECT * FROM customers ORDER BY age ASC", con)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))

sns.barplot(data=df_age, x="age", y="name")
plt.title("Sammy Johnson is the Oldest Customer")
plt.xlabel("Age (years)")
plt.ylabel("Customer Name")
plt.show()
```
