```python
def multiply_for (a,b):
    result = 0
    if b >=0:
        for x in range (1, b+1):
            result = result + a
    else:
        for x in range (1, -b+1):
            result = result - a
    return result 
print (multiply_for(3,5))
```

```python
def multiply_while (a,b):
    result = 0
    x=1
    
    if b >=0:
        while x <= b:
            result= result + a
            x = x +1
    else:
         while x <= -b:
            result= result - a
            x = x +1
    return result
print (multiply_while (3,5)) 
```

```python
print (multiply_for(3,5)) 
print (multiply_while (3,5)) 
3*5
```

The while loop terminates by adding a condition, in this case x <= b,that tells Python to stop the loop after the variable (counter) becomes greater than the endpoint, which is 5 in this problem.
