```python
df.plot.scatter(x="X", y="Y") .set_title("Truck List Price vs.Best Price")
plt.xlabel ("List Price ($1,000s)") 
plt.ylabel ("Best Price ($1,000s)")
```

The Pearson's r between the list price and best price is 0.99854. These results show that they have a very strong positive correlation. The scatter plot shows a very strong positive linear relationship between list price and best price. 

For a list price of 18,000, I predict the best price to be approximately 15,800 based on the near perfect line in the scatter plot. I am confident in this prediction because of the almost perfect correlation of 0.99854 and nearly straight line. Also, 18,000 falls within the prices of this dataset,but the strong correlation doesn't guarantee an exact selling price.
