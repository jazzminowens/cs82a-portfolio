```python
df["cover"]
```

The median of the cover type is paperback (pb). There are 15 books, so the median is the 8th value after sorting which is pb.

```python
print(df["weight"].mean())
```

The mean weight for all books is 706.67 grams. This represents the average weight of the 15 books and is close to the median weight of 725 grams.

```python
print(df.groupby("cover")["weight"].mean())
```

The mean weight for hardback books is 796.43 grams and paperback books is 628.13 grams. Hardback books are 168.3 grams heavier on avergae than paperback books. Based on these results, the library may want to choose paperback books in the future since they weigh less.
