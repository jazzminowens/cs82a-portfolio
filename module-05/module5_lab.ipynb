**Chick Feed Comparison Null Hypothesis**
There is no differnece in the average weight of casein-fed and horsebean-fed chicks.


```python
import pandas as pd
df = pd.read_csv("chickwts.csv")
```

```python
df.head()
```

```python
from scipy.stats import ttest_ind
a = df[df["feed"] == "casein"]["weight"]
b = df[df["feed"] == "horsebean"]["weight"]
ttest_ind(a, b)
```

**Conclusion**


Caesin-fed chicks' average weight was 323.58 g and horsebean-fed chicks' 160.20 g, an effect size of 163.38 gwith a p-value of 8.25e-07, so I reject the null hypothesis of equal average weights, with caution because the data is from only one farm, which limits how these results can be applied.
