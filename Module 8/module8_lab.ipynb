```python

# Import the libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the CSV file
# Replace penguins.csv with your actual file name
df = pd.read_csv("penguins.csv")

# Step 1: Check for missing values in each column
print("Missing Values:")
print(df.isnull().sum())

# Step 2: Compare average bill length across species
# Group the data by species and calculate the mean
print("\nAverage Bill Length by Species:")
print(df.groupby("species")["bill_length_mm"].mean())

# Step 3: Create a boxplot to compare bill lengths
# The x-axis shows species and the y-axis shows bill length
sns.boxplot(data=df, x="species", y="bill_length_mm")

# Add a title and labels to the chart
plt.title("Bill Length Across Penguin Species")
plt.xlabel("Penguin Species")
plt.ylabel("Bill Length (mm)")

# Display the chart
plt.show()
```

## Improved AI Chart


```python

# Step 3: Create a boxplot with three different colors
plt.figure(figsize=(9, 6))

# Assign a different color to each penguin species
colors = ["skyblue", "mediumpurple", "lightgreen"]

sns.boxplot(
    data=df,
    x="species",
    y="bill_length_mm",
    hue="species",
    palette=colors,
    legend=False
)

# Add a descriptive title and axis labels
plt.title("Bill Length Across Penguin Species", fontsize=14)
plt.xlabel("Penguin Species", fontsize=12)
plt.ylabel("Bill Length (mm)", fontsize=12)

# Add horizontal gridlines for easier comparison
plt.grid(axis="y", linestyle="--", alpha=0.3)

# Adjust spacing and display the chart
plt.tight_layout()
plt.show()
```
