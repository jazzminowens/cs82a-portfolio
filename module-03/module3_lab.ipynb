### Intial Obesrvations
Three things that I  see wrong  with this data  set  is: 
1.  The  price count is 288 but there are 300 rows, so there may be missing values.
2. The minimum for quantity is -5 which seems suspicious for a sales order.
3. The date column has inconsistent formats, some follow MM/DD/YYYY and others YYYY-MM-DD.

## Cleaning Log

- There was a total of 300 rows loaded
- Filled 12 missing price values with median 37.53.
- The maximum price is 479.29, the mean is 56.07 and median is 37.53.
  I chose the median because the high price can skew the data and mean.
- Removed 8 duplicate rows leaving 292.
- Converted zip codes from integers to string and restored leading zeroes by adapting to 5-character zips.
- Converted the date column from object to real datetimes and standardized the mixed formats.
- Decided to  treat the 3 negative quantities (-5,-4,-2) as possiblle returns instead of typos. I decided to keep these the rows instead of correcting or deleting, since return of items is possiblele.
- A wrong choice at any of these steps could mislead a business decision becuase it could give a false idea of revenue or even the amount of a product that is sold. For example, if i decided to keep the duplicated rows that would increase the total revenue and product quantity.
