```python

# Step 1: Import the necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Step 2: Load the CSV file
# Replace savings.csv with your actual file name
df = pd.read_csv("lifecyclesavings.csv")

# Step 3: Check the first five rows
print(df.head())

# Step 4: Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Step 5: Select the savings rate column for clustering
X = df[["sr"]]

# Step 6: Use K-Means to divide countries into 3 groups
# random_state makes the results reproducible
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

# Assign each country to a cluster
df["Cluster"] = kmeans.fit_predict(X)

# Step 7: Calculate the average savings rate for each cluster
print("\nAverage Savings Rate by Cluster:")
print(df.groupby("Cluster")["sr"].mean())

# Step 8: Create a boxplot comparing savings rates
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df,
    x="Cluster",
    y="sr",
    hue="Cluster",
    palette=["skyblue", "mediumpurple", "lightgreen"],
    legend=False
)

# Step 9: Add a title and axis labels
plt.title("Savings Rate by Country Cluster", fontsize=14)
plt.xlabel("Country Cluster", fontsize=12)
plt.ylabel("Savings Rate (%)", fontsize=12)

# Add gridlines for easier comparison
plt.grid(axis="y", linestyle="--", alpha=0.3)

# Step 10: Display the chart
plt.tight_layout()
plt.show()
```
