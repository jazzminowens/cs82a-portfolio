**Interpertation:** After cleaning the data income and number of vehicles had the highest correlation with a Pearson's r value of 0.134. Age and income had almost no linear relationship at all with a r value of 0.009160 and age and number of vehicles had a very weak negative relationship with a r-value of -0.069.

**Cleaning Log**

- There was a total of 1000 rows loaded. I decided to drop the 56 rows which had data missing for num.vehicles so all 3 correlations used complete rows.
- I excluded 1 row that had a negative income as negative income in this case is invalid.
- I excluded 3 rows that had an age of 0 because this is unrealistic as they are also listed as married and divorced. 
- I excluded 7 rows that had an age greater than 115 as these ages seem unrealistic. I chose 115 as my upper age cutoff for this analysis. 
- In total 67 rows were excluded from the analysis.
