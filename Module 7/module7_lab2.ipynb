```python
import sklearn
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
df = pd.read_csv("ratings.tsv", sep="\t")
```

```python
df.head()
```

```python
X = df[["ambience", "food", "service"]]
y = df["rating"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
len(X_train)
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train, y_train)
print(model.score(X_test, y_test))
```

```python
print(model.coef_, model.intercept_)
```

**Ranking of coefficents:** Food(0.9568), Service(0.2038), Ambience(0.0740)

**Recommendation:** I recommend this business to focus on fixing the food first to increase their overall rating. This is because food had the highest coefficent out of the 3 with 0.9568. This coefficent predicts a 0.96 point increase for each 1 point increase in the overall rating, holding ambience and service constant.
