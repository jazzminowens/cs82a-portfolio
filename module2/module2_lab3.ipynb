The temperature cutoffs are:
- Cold if the temperature is below 15°C
- Warm if the temperature is between 15°C and 25°C
- Hot if the temperature is above 25°C
- Here is the list of temperatures:

```python
temps = [10, 16, 23, 28, 14, 22, 30]
```

```python
for t in temps:
    if t> 25:
        print (t, "Hot")
    elif t>= 15:
        print (t, "Warm")
    else: 
        print (t, "Cold")
```
